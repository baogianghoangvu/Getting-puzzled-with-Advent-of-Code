```python
import itertools
import string
from pathlib import Path

import numpy as np
```

```python
puzz = """
30373
25512
65332
33549
35390
""".strip().split(
    "\n"
)
puzz = Path("./day8.txt").read_text().splitlines()
```

## Part 1

```python
puzz = np.array([[int(i) for i in l] for l in puzz])
m, n = puzz.shape

visi = np.array(
    [
        (
            [1 for _ in range(n)]
            if _ in (0, m - 1)
            else [1 if _ in (0, n - 1) else 0 for _ in range(n)]
        )
        for _ in range(m)
    ]
)
```

```python
for i, j in itertools.product(range(1, m - 1), range(1, n - 1)):
    abov = puzz[:i, j]
    belo = puzz[i + 1 : m, j]
    left = puzz[i, :j]
    righ = puzz[i, j + 1 : n]
    isviz = int(puzz[i, j] > min(abov.max(), belo.max(), left.max(), righ.max()))
    visi[i, j] = isviz
```

## Part 2

```python
scor = np.zeros_like(visi)
```

```python
for i, j in itertools.product(range(1, m - 1), range(1, n - 1)):
    abov = puzz[:i, j][::-1]
    belo = puzz[i + 1 : m, j]
    left = puzz[i, :j][::-1]
    righ = puzz[i, j + 1 : n]

    center = puzz[i, j]

    a = np.where((center > abov) == False)[0]
    b = np.where((center > belo) == False)[0]
    l = np.where((center > left) == False)[0]
    r = np.where((center > righ) == False)[0]

    view_a = abov.size if a.size == 0 else 1 + a[0]
    view_b = belo.size if b.size == 0 else 1 + b[0]
    view_l = left.size if l.size == 0 else 1 + l[0]
    view_r = righ.size if r.size == 0 else 1 + r[0]

    scor[i, j] = view_a * view_b * view_l * view_r
```
