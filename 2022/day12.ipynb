## Part 1

```python
matrix = np.array([[ord(c) - ord("a") for c in p] for p in puzz])

m, n = matrix.shape

start = tuple(int(i[0]) for i in np.array(np.where(matrix == ord("S") - ord("a"))))
end = tuple(int(i[0]) for i in np.array(np.where(matrix == ord("E") - ord("a"))))

matrix[start] = 0
matrix[end] = 25
```

```python
G = nx.DiGraph()
for i, j in itertools.product(range(m), range(n)):
    center = (i, j)

    check_list = []
    if 0 < i:
        check_list.append((i - 1, j))
    if i < m - 1:
        check_list.append((i + 1, j))
    if 0 < j:
        check_list.append((i, j - 1))
    if j < n - 1:
        check_list.append((i, j + 1))

    for coord in check_list:
        if matrix[coord] <= matrix[center] + 1:
            G.add_edge(center, coord)

len(nx.shortest_path(G, source=start, target=end)) - 1
```

```python
G = nx.DiGraph()
for i, j in itertools.product(range(m), range(n)):
    center = (i, j)

    check_list = []
    if 0 < i:
        check_list.append((i - 1, j))
    if i < m - 1:
        check_list.append((i + 1, j))
    if 0 < j:
        check_list.append((i, j - 1))
    if j < n - 1:
        check_list.append((i, j + 1))

    for coord in check_list:
        if 0 <= matrix[coord] - matrix[center] <= 1:
            G.add_edge(center, coord)

len(nx.shortest_path(G, source=start, target=end)) - 1
```

```python
visit = [start]
queue = [dict(coord=start, hist=[start])]

while queue:
    center, history = queue.pop(0).values()
    visit.append(center)

    check_list = []
    i, j = center
    if 0 < i:
        check_list.append((i - 1, j))
    if i < m - 1:
        check_list.append((i + 1, j))
    if 0 < j:
        check_list.append((i, j - 1))
    if j < n - 1:
        check_list.append((i, j + 1))

    for coord in check_list:
        if coord in visit:
            continue
        elif matrix[coord] <= matrix[center] + 1:
            visit.append(coord)
            d = dict(coord=coord, hist=history + [coord])
            queue.append(d)
            if coord == end:
                reach = d
```

```python
len(reach["hist"]) - 1
```

## Part 2

```python
paths = dict()
for coord in zip(*np.where(matrix == 0)):
    coord = tuple(int(i) for i in coord)
    try:
        paths[coord] = len(nx.shortest_path(G, source=coord, target=end)) - 1
    except nx.NetworkXNoPath:
        paths[coord] = math.inf
```
