```python
from pathlib import Path
```

```python
puzz = """
root: pppw + sjmn
dbpl: 5
cczh: sllz + lgvd
zczc: 2
ptdq: humn - dvpt
dvpt: 3
lfqf: 4
humn: 5
ljgn: 2
sjmn: drzm * dbpl
sllz: 4
pppw: cczh / lfqf
lgvd: ljgn * ptdq
drzm: hmdt - zczc
hmdt: 32
""".strip().split(
    "\n"
)
puzz = Path("./day21.txt").read_text().splitlines()
```

```python
puzz[:5]
```

## Part 1

```python
monkay = dict()

for l in puzz:
    contents = l.split()
    if len(contents) == 2:
        key, val = contents
        key = key[:-1]
        val = val
        monkay[key] = {"val": val}
    else:
        key, m1, op, m2 = contents
        key = key[:-1]
        monkay[key] = {"val": None, "m1": m1, "m2": m2, "op": op}
```

```python
tudu = ["root"]
while tudu:
    target = monkay[tudu[-1]]
    if target["val"] is not None:
        tudu.pop()
    else:
        m1 = monkay[target["m1"]]
        m2 = monkay[target["m2"]]

        if m1["val"] is not None and m2["val"] is not None:
            target["val"] = str(int(eval(m1["val"] + target["op"] + m2["val"])))
            tudu.pop()
            continue

        if m1["val"] is None:
            tudu.append(target["m1"])
        if m2["val"] is None:
            tudu.append(target["m2"])
```

## Part 2

```python
from sympy import solve

monke = dict()
for line in puzz:
    m, yell = line.split(': ')
    if ' ' not in yell:
        monke[m] = {'number': True, 'yell': yell}
    else:
        m1, op, m2 = yell.split(' ')
        monke[m] = {'number': False, 'monkes': [m1,m2], 'yell': yell, 'operation':op}


start = "root"
equation = monke[start]["yell"].replace(monke[start]["operation"], "=")
queue = [start]
humn = "humn"

while queue:
    m = queue.pop(0)
    if m == humn:
        continue
    equation = equation.replace(m, f"({monke[m]['yell']})")
    if not monke[m]["number"]:
        next_monkeys = monke[m]["monkes"]
        for n in next_monkeys:
            queue.append(n)

left, right = equation.split("=")
if humn in left:
    num = int(eval(right))
    assess = left
else:
    num = int(eval(left))
    assess = right

assess += f"- {num}"
res = solve(assess, humn)

int(res[0])
```
