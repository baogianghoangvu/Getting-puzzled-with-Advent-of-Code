# https://adventofcode.com/2022

```python
import string
from pathlib import Path
```

```python
puzz = Path("./day5.txt").read_text().splitlines()
```

```python
stack_ = puzz[: puzz.index("") - 1]
moves_ = puzz[puzz.index("") + 1 :]
```

## Part 1

```python
stack = dict()
for s in stack_[::-1]:
    i = 1
    while 1 + 4 * (i - 1) < len(s):
        crate = s[1 + 4 * (i - 1)]
        if crate != " ":
            if i in stack:
                stack[i].append(crate)
            else:
                stack[i] = [crate]
        i += 1
stack
```

```python
for m in moves_:
    _, n, _, fr, _, to = m.split(" ")
    n, fr, to = int(n), int(fr), int(to)
    for _ in range(n):
        stack[to].append(stack[fr].pop())
```

```python
"".join([v[-1] for v in stack.values()])
```

## Part 2

```python
stack = dict()
for s in stack_[::-1]:
    i = 1
    while 1 + 4 * (i - 1) < len(s):
        crate = s[1 + 4 * (i - 1)]
        if crate != " ":
            if i in stack:
                stack[i].append(crate)
            else:
                stack[i] = [crate]
        i += 1
stack
```

```python
for m in moves_:
    _, n, _, fr, _, to = m.split(" ")
    n, fr, to = int(n), int(fr), int(to)
    stack[to].extend(stack[fr][-n:])
    for _ in range(n):
        stack[fr].pop()
```

```python
"".join([v[-1] for v in stack.values()])
```
