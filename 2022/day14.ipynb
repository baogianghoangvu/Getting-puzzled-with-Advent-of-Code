```python
from pathlib import Path
```

```python
puzz = """
498,4 -> 498,6 -> 496,6
503,4 -> 502,4 -> 502,9 -> 494,9
""".strip().split(
    "\n"
)
puzz = Path("./day14.txt").read_text().splitlines()
```

```python
puzz[:5]
```

## Part 1

```python
rocks = dict()
for p in puzz[:]:
    coords = [list(map(int, path.split(","))) for path in p.split(" -> ")]
    for i in range(len(coords) - 1):
        x0, y0 = coords[i]
        x1, y1 = coords[i + 1]
        for x in range(min(x0, x1), max(x0, x1) + 1):
            rocks[x] = rocks.get(x, set())
            for y in range(min(y0, y1), max(y0, y1) + 1):
                rocks[x].add(y)

n_rocks = sum([len(v) for v in rocks.values()])


xsad = 500
hellno = True
while hellno:
    xr = xsad
    yr = min(rocks[xr])

    while hellno:
        if xr - 1 not in rocks or xr + 1 not in rocks:
            hellno = False
            break

        sw = yr not in rocks[xr - 1]  # left-down empty
        se = yr not in rocks[xr + 1]  # right-down empty
        if sw:
            xr = xr - 1
        elif se:
            xr = xr + 1
        else:
            rocks[xr].add(yr - 1)
            break

        limbo = [y for y in rocks[xr] if y > yr]
        if limbo:
            yr = min(limbo)
        else:
            hellno = False
            break

sum([len(v) for v in rocks.values()]) - n_rocks
```

## Part 2

```python
rocks = dict()
for p in puzz[:]:
    coords = [list(map(int, path.split(","))) for path in p.split(" -> ")]
    for i in range(len(coords) - 1):
        x0, y0 = coords[i]
        x1, y1 = coords[i + 1]
        for x in range(min(x0, x1), max(x0, x1) + 1):
            rocks[x] = rocks.get(x, set())
            for y in range(min(y0, y1), max(y0, y1) + 1):
                rocks[x].add(y)

n_rocks = sum([len(v) for v in rocks.values()])

yma = 2 + max([max(v) for v in rocks.values()])
xsad = 500
hellno = True
while hellno:
    xr = xsad
    yr = min(*rocks[xr], yma)
    if yr == 0:
        hellno = False
        break

    while hellno:
        if xr - 1 not in rocks:
            rocks[xr - 1] = set([yma])
        if xr + 1 not in rocks:
            rocks[xr + 1] = set([yma])

        sw = yr not in rocks[xr - 1]  # left-down empty
        se = yr not in rocks[xr + 1]  # right-down empty
        if sw:
            xr = xr - 1
        elif se:
            xr = xr + 1
        else:
            rocks[xr].add(yr - 1)
            break

        limbo = [y for y in rocks[xr] if y > yr]
        if limbo:
            yr = min(limbo)
        else:
            yr = yma
            rocks[xr].add(yr)

sum([len(v) for v in rocks.values()]) - n_rocks - len(rocks)
```
