## Part 1

```python
class Tile:
    def __init__(
        self,
        coord: tuple,
        m: int,
        n: int,
        val: str,
    ):
        self.coord = coord
        i, j = coord
        self.up = (i - 1 if i - 1 >= 0 else m - 1, j)
        self.do = (i + 1 if i + 1 < m else 0, j)
        self.le = (i, j - 1 if j - 1 >= 0 else n - 1)
        self.ri = (i, j + 1 if j + 1 < n else 0)
        self.move = {
            0: self.ri,
            1: self.do,
            2: self.le,
            3: self.up,
        }
        self.val = val

    def __repr__(self):
        string = f"{{{self.coord}: `{self.val}`; ^{self.up}^; v{self.do}v; <{self.le}<; >{self.ri}> }}"
        return string


mapp = puzz[:-2]
m = len(mapp)
n = max(len(line) for line in mapp)
mapp = [line.ljust(n) for line in mapp]

graph = dict()
for i in range(m):
    row = mapp[i]
    for j in range(n):
        val = row[j]
        graph[(i, j)] = Tile((i, j), m, n, val)
```

```python
def parse_paths(raw):
    cunu = ""
    paths = []
    for c in raw:
        if c.isdigit():
            cunu += c
        else:
            paths.append(int(cunu))
            paths.append(c)
            cunu = ""
    else:
        paths.append(int(cunu))
    return paths


paths = parse_paths(puzz[-1])
```

```python
paths[-11:]
```

```python
start = (0, mapp[0].index("."))
orie = 0
X, Y = start


for step in paths:
    # print("=" * 55)
    # print(X, Y, step)
    if step == "L":
        orie = (orie - 1) % 4
    elif step == "R":
        orie = (orie + 1) % 4
    elif isinstance(step, int):
        xx, yy = X, Y
        while step > 0:
            xx, yy = graph[(xx, yy)].move[orie]
            nextval = graph[(xx, yy)].val
            if nextval == "#":
                # print(f"{X:2}", f"{Y:2}", "#")
                break
            elif nextval == " ":
                continue
            elif nextval == ".":
                X, Y = xx, yy
                step -= 1
                # print(f"{X:2}", f"{Y:2}", ".")


print((X, Y), orie, 1000 * (X + 1) + 4 * (Y + 1) + orie)
```
