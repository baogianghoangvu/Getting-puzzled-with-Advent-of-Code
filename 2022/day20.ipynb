# https://adventofcode.com/2022

```python
from collections import Counter
from pathlib import Path
```

```python
puzz = """
1
2
-3
3
-2
0
4
""".strip().split(
    "\n"
)
puzz = Path("./day20.txt").read_text().splitlines()
```

```python
puzz[:5]
```

## Part 1

```python
ops = {i: int(p) for i, p in enumerate(puzz)}
sequence = list(ops.keys())
n = len(sequence)

# print(f"  ", [ops[s] for s in sequence])
for item, op in ops.items():

    idx_cur = sequence.index(item)
    idx_new = (idx_cur + op - 1) % (n - 1) + 1

    sequence.insert(idx_new, sequence.pop(idx_cur))
    # print(f"{op:2}", [ops[s] for s in sequence])


whereiszego = [k for k, v in ops.items() if v == 0][0]

res = []
for th in [1000, 2000, 3000]:
    idx = th + sequence.index(whereiszego)
    idx %= n
    res.append(ops[sequence[idx]])
    print(th, idx, ops[sequence[idx]])
sum(res), res
```
