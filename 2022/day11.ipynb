```python
import copy
import math
import string
from collections import deque
from pathlib import Path
```

```python
puzz = """
Monkey 0:
  Starting items: 79, 98
  Operation: new = old * 19
  Test: divisible by 23
    If true: throw to monkey 2
    If false: throw to monkey 3

Monkey 1:
  Starting items: 54, 65, 75, 74
  Operation: new = old + 6
  Test: divisible by 19
    If true: throw to monkey 2
    If false: throw to monkey 0

Monkey 2:
  Starting items: 79, 60, 97
  Operation: new = old * old
  Test: divisible by 13
    If true: throw to monkey 1
    If false: throw to monkey 3

Monkey 3:
  Starting items: 74
  Operation: new = old + 3
  Test: divisible by 17
    If true: throw to monkey 0
    If false: throw to monkey 1
""".strip().split(
    "\n"
)
puzz = Path("./day11.txt").read_text().splitlines()
```

## Part 1

```python
tracker_og = dict()

monkey_idx = 0
for p in puzz:
    if p.startswith("Monkey "):
        tracker_og[monkey_idx] = dict(
            items=deque(),
            op="",
            test=0,
            true=0,
            false=0,
            inspect=0,
        )

    elif p.startswith("  Starting items: "):
        tracker_og[monkey_idx]["items"] = deque([int(i) for i in p[18:].split(", ")])
    elif p.startswith("  Operation: new = "):
        tracker_og[monkey_idx]["op"] = p[19:]

    elif p.startswith("  Test: divisible by "):
        tracker_og[monkey_idx]["test"] = int(p[21:])
    elif p.startswith("    If true: throw to monkey "):
        tracker_og[monkey_idx]["true"] = int(p[-1])
    elif p.startswith("    If false: throw to monkey "):
        tracker_og[monkey_idx]["false"] = int(p[-1])

    elif p == "":
        monkey_idx += 1

    else:
        raise ("wtf")
```

```python
tracker = copy.deepcopy(tracker_og)

for i in range(20):
    for mi, meta in tracker.items():
        tracker[mi]["inspect"] += len(meta["items"])
        while meta["items"]:
            op = lambda old: eval(meta["op"])

            old = meta["items"].popleft()
            new = op(old) // 3

            test = (new % meta["test"]) == 0
            target = meta["true"] if test else meta["false"]

            tracker[target]["items"].append(new)
```

```python
inspects = [v["inspect"] for k, v in tracker.items()]
inspects.sort()
inspects[-1] * inspects[-2]
```

## Part 2

```python
tracker = copy.deepcopy(tracker_og)

lcm = math.lcm(*[v["test"] for k, v in tracker.items()])

for i in range(10_000):
    if (i + 1) % 1000 == 0:
        print(i + 1)
    for mi, meta in tracker.items():
        tracker[mi]["inspect"] += len(meta["items"])
        while meta["items"]:
            op = lambda old: eval(meta["op"])

            old = meta["items"].popleft()
            new = op(old) % lcm

            test = (new % meta["test"]) == 0
            target = meta["true"] if test else meta["false"]

            tracker[target]["items"].append(new)
```

```python
inspects = [v["inspect"] for k, v in tracker.items()]
inspects.sort()
inspects[-1] * inspects[-2]
```
