```python
snadec = {"2": 2, "1": 1, "0": 0, "-": -1, "=": -2}
decsna = {v: k for k, v in snadec.items()}


def ufans(s):
    n = 0
    for i, c in enumerate(s[::-1], 0):
        match c:
            case "=":
                normie = -2
            case "-":
                normie = -1
            case "0":
                normie = 0
            case "1":
                normie = 1
            case "2":
                normie = 2
        n += normie * 5**i
    return n


def snafu(n):
    s = ""
    rem = 0
    while n > 0:
        digit = n % 5 + rem
        rem = 0
        if digit > 2:
            rem = 1
            digit -= 5
        else:
            rem = 0
        s = str(decsna[(digit)]) + s
        n //= 5
    if rem == 1:
        s = "1" + s
    s or "0"
    return s
```

```python
todec = 0
for s in puzz:
    todec += ufans(s)

snafu(todec)
```
