```python
import string
from pathlib import Path
```

```python
puzz = """noop
addx 3
addx -5
""".strip().split(
    "\n"
)

puzz = Path("./day10.txt").read_text().splitlines()
```

```python
puzz[:5]
```

## Part 1

```python
cycles = [1]

for p in puzz:
    if p == "noop":
        cycles.append(cycles[-1])
    elif p.startswith("addx "):
        n = int(p.split(" ")[1])
        cycles.append(cycles[-1])
        cycles.append(cycles[-1] + n)
```

```python
interestanh = [(20 + i * 40) for i in range(6)]
interestanh
```

```python
sum([i * cycles[i - 1] for i in interestanh])
```

## Part 2

```python
drawing_flat = []
for crt, sprite in enumerate(cycles):
    rem = crt % 40
    if rem:
        crt = rem
    else:
        crt = 40
    if sprite - 1 <= crt <= sprite + 1:
        drawing_flat.append("#")
    else:
        drawing_flat.append(" ")
```

```python
drawing_2d = []
for i in range(6):
    drawing_2d.append(drawing_flat[i * 40 : (i + 1) * 40])

for l in drawing_2d:
    print(" ".join(l))
```
