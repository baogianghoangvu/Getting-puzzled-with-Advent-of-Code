```python
def viz(elves):
    xmi = +float("inf")
    xma = -float("inf")
    ymi = +float("inf")
    yma = -float("inf")

    for i in elves.values():
        xmi = int(min(i.real, xmi))
        xma = int(max(i.real, xma))
        ymi = int(min(i.imag, ymi))
        yma = int(max(i.imag, yma))

    viz = [["  " for _ in range(yma - ymi + 1)] for _ in range(xma - xmi + 1)]
    for cccd, coord in elves.items():
        x = int(coord.real) - xmi
        y = int(coord.imag) - ymi
        viz[x][y] = f"{cccd:2}"

    return np.array(viz)
```
