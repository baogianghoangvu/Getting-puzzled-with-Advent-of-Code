```python
import string
from pathlib import Path
```

```python
puzz = Path("./day9.txt").read_text().splitlines()
```

## Part 1

```python
def head_move(hp, d):
    x, y = hp
    if d == "U":
        x -= 1
    elif d == "D":
        x += 1
    elif d == "L":
        y -= 1
    elif d == "R":
        y += 1
    return (x, y)


def tail_move(tp, hp):
    x, y = tp
    hx, hy = hp
    dist_x = hx - x
    dist_y = hy - y

    if (-1 <= dist_x <= 1) and (-1 <= dist_y <= 1):
        pass
    else:
        x += 1 if (dist_x > 0) else -1 if (dist_x < 0) else 0
        y += 1 if (dist_y > 0) else -1 if (dist_y < 0) else 0

    return (x, y)
```

```python
hp = tp = (0, 0)

visited = set()
for p in puzz:
    d, n = p.split(" ")
    n = int(n)
    for _ in range(n):
        hp = head_move(hp, d)
        tp = tail_move(tp, hp)
        visited.add(tp)

len(visited)
```
