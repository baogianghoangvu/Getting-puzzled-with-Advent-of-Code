```python
import string
from pathlib import Path
```

```python
puzz = """
[1,1,3,1,1]
[1,1,5,1,1]

[[1],[2,3,4]]
[[1],4]

[9]
[[8,7,6]]

[[4,4],4,4]
[[4,4],4,4,4]

[7,7,7,7]
[7,7,7]

[]
[3]

[[[]]]
[[]]

[1,[2,[3,[4,[5,6,7]]]],8,9]
[1,[2,[3,[4,[5,6,0]]]],8,9]
""".strip().split(
    "\n"
)
puzz = Path("./day13.txt").read_text().splitlines()
```

## Part 1

```python
idx = 1
pairs = {idx: []}

for p in puzz:
    if p.startswith("["):
        pairs[idx].append(eval(p))
    else:
        idx += 1
        pairs[idx] = []
```

```python
def recu(left, right):
    int_left = isinstance(left, int)
    int_right = isinstance(right, int)

    if int_left and int_right:
        if left < right:
            return True
        elif left > right:
            return False

    elif not int_left and not int_right:
        left = left.copy()
        right = right.copy()
        while left and right:
            res = recu(left.pop(0), right.pop(0))
            if res is not None:
                return res

        if left:
            return False
        elif right:
            return True

    else:
        left = [left] if int_left else left.copy()
        right = [right] if int_right else right.copy()
        return recu(left, right)
```

```python
valid = []
for i, pair in pairs.items():
    if recu(*pair):
        valid.append(i)
sum(valid)
```

## Part 2

```python
packets = {
    1: [[2]],
    2: [[6]],
}
idx = 3
for p in puzz:
    if p.startswith("["):
        packets[idx] = eval(p)
        idx += 1
```

```python
for i in range(1, len(packets)):
    while i >= 1:
        if recu(packets[i], packets[i + 1]):
            break
        else:
            packets[i], packets[i + 1] = packets[i + 1], packets[i]
            i -= 1
```

```python
res = [k for k, v in packets.items() if str(v) in ["[[2]]", "[[6]]"]]
```

```python
res[0] * res[1]
```
